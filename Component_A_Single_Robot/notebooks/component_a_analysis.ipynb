```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# We need to load trajectory log generated by Webots
df = pd.read_csv(r'C:\\Users\\User\\OneDrive\\Desktop\\2026_SecondSemester\\Assignments\\robotics_assignment\\component_a_single_robot\\controllers\\a_robot_controller\\trajectory_log.csv')

#We then move to set the publication quality styling
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5), dpi=300)

# Plot 1: 2D Spatial Trajectory (Planned vs Actual)
ax1.plot(df['x_actual'], df['y_actual'], label='Actual Trajectory (GPS)', color='#1f77b4', linewidth=2)
ax1.plot(df['x_target'], df['y_target'], 'r--', label='Target Waypoints', linewidth=1.5, marker='o')
ax1.set_title('Component A: Closed-Loop Trajectory Tracking', fontsize=12, fontweight='bold')
ax1.set_xlabel('X Position (m)')
ax1.set_ylabel('Y Position (m)')
ax1.legend(loc='best')
ax1.grid(True)

# Plot 2: PID Heading Error Convergence over Time
ax2.plot(df['time'], np.degrees(df['heading_error']), color='#d62728', linewidth=1.5)
ax2.set_title('PID Heading Error Convergence', fontsize=12, fontweight='bold')
ax2.set_xlabel('Time (s)')
ax2.set_ylabel('Heading Error (degrees)')
ax2.grid(True)

plt.tight_layout()
plt.show()
```

```python
#Saving the figure as a high-resolution PNG file
plt.tight_layout()
plt.savefig('component_a_tracking_results.png', dpi=300, bbox_inches='tight')
plt.show()
```
